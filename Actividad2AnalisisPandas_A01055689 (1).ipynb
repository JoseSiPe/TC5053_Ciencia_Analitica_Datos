8. Utiliza el dataframe de indicadores diarios para crear un nuevo dataframe `indicadores_mensuales`. Para ello:
- Crea una nueva columna `mes` a partir de la fecha que se encuentra en el índice.
- Haciendo uso de `groupby()`, calcula los promedios mensuales y conserva únicamente las columnas `humedad_relativa`, `temperatura_celsius` y `velocidad_viento` en el nuevo dataframe.

```python
orden_meses = ["January", "February", "March", "April", "May", "June",
               "July", "August", "September", "October", "November", "December"]

indicadores_diarios["mes"] = pd.Categorical(
    pd.to_datetime(indicadores_diarios.index).month_name(),
    categories=orden_meses,
    ordered=True
)

indicadores_mensuales = indicadores_diarios.groupby("mes", observed=True)[
    ["humedad_relativa", "temperatura_celsius", "velocidad_viento"]
].mean()

print(indicadores_mensuales)
```

9. Construye el mismo dataframe mensual, pero esta vez utilizando la función `pivot_table()`.
- Añade la columna `indice_calor` usando la fórmula de Thom (*Temperature - Humidity Index*, TDI):
> `temperatura_celsius + 0.33 x humedad_relativa - 0.7 x velocidad_viento - 4`

```python
indicadores_mensuales_pivot = pd.pivot_table(
    indicadores_diarios,
    index="mes",
    values=["humedad_relativa", "temperatura_celsius", "velocidad_viento"],
    aggfunc="mean",
    observed=True
)

indicadores_mensuales_pivot["indice_calor"] = (
    indicadores_mensuales_pivot["temperatura_celsius"]
    + 0.33 * indicadores_mensuales_pivot["humedad_relativa"]
    - 0.7 * indicadores_mensuales_pivot["velocidad_viento"]
    - 4
)

print(indicadores_mensuales_pivot)
```

10. Recuerda que de las cuatro funciones estudiadas para manipular la estructura del dataframe:
- `melt/pivot` hacen las transformaciones de manera controlada, es decir puedes
definir por medio de sus parámetros qué variables quedarán como índices, columnas y valores.
- `stack/unstack` la conversión se aplica siempre sobre los niveles inferiores de index/columns.
- Convierte el dataframe del ejercicio anterior a formato largo usando: `melt()` y `stack()` y comenta las diferencias.

```python
largo_melt = indicadores_mensuales_pivot.reset_index().melt(id_vars="mes", var_name="indicador", value_name="valor")
print(largo_melt)

largo_stack = indicadores_mensuales_pivot.stack()
print(largo_stack)
```

Diferencias entre `melt()` y `stack()`

- **`melt()`** requiere pasar `mes` del índice a columna con `reset_index()` y permite controlar la estructura: qué columna se conserva como identificador (`id_vars`) y cómo se llaman las nuevas columnas (`var_name`, `value_name`). Devuelve un **DataFrame** de 48 filas con índice numérico, ordenado por indicador (primero los 12 meses de humedad, luego los de temperatura, etc.).
- **`stack()`** no necesita parámetros: toma el nivel inferior de las columnas y lo convierte en un nivel del índice. Devuelve una **Serie con MultiIndex** (`mes`, indicador), ordenada por mes (los 4 indicadores de enero, luego los de febrero, etc.), y el nuevo nivel del índice y los valores quedan sin nombre.
- En resumen: ambos producen los mismos 48 valores en formato largo; `melt` ofrece más control y un resultado tabular listo para usarse, mientras que `stack` es más directo cuando el índice ya tiene significado y se quiere trabajar con índices jerárquicos.
